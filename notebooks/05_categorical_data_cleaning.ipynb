## Scope and Objective

This notebook focuses on inspecting and cleaning categorical features to ensure
consistency and robustness for downstream encoding and modeling steps.

The analysis is limited to category standardization and rare category inspection.
No exploratory analysis, feature–target evaluation, or modeling is performed.


```python
import sys
from pathlib import Path
import pandas as pd

# Add project root to PYTHONPATH
project_root = Path().resolve().parent
sys.path.append(str(project_root))

from src.utils.config import PROCESSED_DATA_PATH
```

```python
df = pd.read_csv(PROCESSED_DATA_PATH / "telco_customer_churn_validated.csv")
```

## Identify Categorical Features

```python
categorical_cols = df.select_dtypes(include=["object"]).columns.tolist()
categorical_cols.append("SeniorCitizen")
categorical_cols.remove("TotalCharges")
categorical_cols.remove("customerID")
categorical_cols.remove("Churn")
categorical_cols
```

## Category Consistency Checks

```python
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].unique())
```

### Category Consistency Checks - Key Findings

- All categorical values were inspected for inconsistencies such as typos,
unexpected casing, or formatting variations. No inconsistencies were found,
and all categorical features exhibit consistent and well-defined values.

## Rare Category Inspection

```python
threshold = 0.01

for col in categorical_cols:
    print(f"\nAnalyzing column: {col}")
    value_counts = df[col].value_counts(normalize=True)
    rare_categories = value_counts[value_counts < threshold].index.tolist()

    if rare_categories:
        print(
            f"Rare categories in '{col}' (less than {threshold*100}%): {rare_categories}"
        )
    else:
        print(f"No rare categories in '{col}'")
```

### Rare Category Inspection – Key Findings

- A rare category inspection was performed using a 1% frequency threshold.
No categorical levels were found below this threshold, indicating the absence
of rare categories that would require merging or removal.

## Cleaning Decisions
- As no data quality issues were detected, no categorical values should be modified
  or removed at this stage.
