# Duplicate and Inconsistent row check

## Scope and Objective

This notebook validates the dataset by inspecting duplicate records and

to modeling.

No exploratory analysis or feature–target evaluation is performed.

## Logical Consistency Checks

```python
df[(df["PhoneService"] == "No") & (df["MultipleLines"] != "No phone service")].shape[0]
```

```python
internet_services = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
]

for col in internet_services:
    inconsistent = df[
        (df["InternetService"] == "No") & (df[col] != "No internet service")
    ]
    print(f"{col}: {inconsistent.shape[0]}")
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors="coerce").fillna(0)
df[(df["tenure"] == 0) & (df["TotalCharges"] != 0)].shape[0]
```

### Logical Consistency Checks - key findings
- No logically inconsistent rows were identified across service-related
  features.
